# **DATA PREPROCESSING**

## **Data Cleansing**

```python
# Identify the number of missing values
nullValues_columns = df_creditCards.isnull().sum()
nullValues_rows = df_creditCards.isnull().sum(axis=1)
print(nullValues_columns)
print(nullValues_rows)
print("\nTotal number of null values =", nullValues_columns.sum())
```

There are no missing values (NaN) in the dataset, therefore removing/imputing is not required.

```python
# Eliminate duplicate rows directly on the DataFrame
df_creditCards.drop_duplicates(keep='first', inplace=True)
df_creditCards.shape
```

No need to check for ID uniqueness since it is not a provided feature, instead automatically assigned when dataset is read.

To handle potentially misfiled values or values of inconsistent format, we need to have knowledge of the range of values and the particular problem domain, which is not possible due to columns being anonymous.

```python
# To handle invalid values, we can try to convert them to numerical values (e.g.
# "1.045" to 1.045), and replace with NaN if it is not a numerical value.
for column in df_creditCards.columns:
    df_creditCards[column] = pd.to_numeric(df_creditCards[column], errors='coerce')
```

```python
# RESULTING DATASET AFTER DATA CLEANSING
df_creditCards.info()
```

## Data Reduction

Due to the fact that components are obtained with PCA transformation, the data is completely anonymous. Therefore, it is not possible to understand whether the features or data is necessary or not. Eliminating rows/columns arbitrarily may lead to unintended effects in the model's prediction capability, and misleading results. Hence, the data reduction step is skipped.

## Data Scaling

Values of the "Amount" column doesn't align with the rest of the values. Therefore we will rescale these values using Normalization to a range of (-1,1).

```python
from sklearn.preprocessing import StandardScaler

df_creditCards['Amount_Normalized'] = StandardScaler().fit_transform(df_creditCards['Amount'].values.reshape(-1, 1))
df_creditCards.drop(['Amount'], axis=1, inplace=True)
df_creditCards.head()
```

Values of the "Time" and "Class" columns can be left as-is, since the former can be used to plot readable graphs and the latter will be used as the "label".

Since all the values are known to be numerical, there is no need to check data type compatibility between different data types.

A look into the distribution of each feature.

```python
def draw_histograms(dataframe, features, rows, cols):
    fig=plt.figure(figsize=(20,20))
    for i, feature in enumerate(features):
        ax=fig.add_subplot(rows,cols,i+1)
        dataframe[feature].hist(bins=20,ax=ax,facecolor='blue')
        ax.set_title(feature+" Distribution",color='black')
        ax.set_yscale('log')
    fig.tight_layout()
    plt.show()
draw_histograms(df_creditCards,df_creditCards.columns,8,4)
```

Since the dataset is known to be unbalanced, a closer look into the distribution of the labels may prove useful (class=0=non-fraud, class=1=fraud).

```python
plot = sns.countplot(x='Class', data=df_creditCards)
plt.ylabel('Number of Transactions')
plt.title('Fraudulent vs Non-Fraudulent Transactions')
plot.set_yscale('log') #setting the y-axis to logarithmic scale to make
# visualisation easier, due to the imbalance in the dataset
plt.show()
df_creditCards.Class.value_counts()
```

In the dataset there is an imbalance between the number of fraudulent and non-fradulent transactions, in other words there are a very small number of fraudulent transactions. Therefore, if we plot the variation with time considering the two cases together on the same grid, the graph won't give much visual information.

```python
sns.kdeplot(data=df_creditCards, x='Time', hue='Class', common_grid=True, clip=(0, None), palette=['green', 'red'], hue_order=[0,1])
plt.legend(labels=['Fraud', 'Non-Fraud'])
plt.xlabel('Time')
```

That is why, instead we just consider both cases individually. Even though it may seem like there are a large number of frauds, this graph only demonstrates the density of occurence for each case relative to time.

Plotting the variations in transactions relative to time can help uncover temporal patterns in fraudulent activity, anomalies and outliers (e.g. unusual spikes, time gaps), and idenfity the overall behavior of each type of transaction.

```python
sns.kdeplot(data=df_creditCards.loc[df_creditCards['Class']==1]['Time'], label='Fraud', clip=(0, None), color='red')
sns.kdeplot(data=df_creditCards.loc[df_creditCards['Class']==0]['Time'], label='Non-Fraud', clip=(0, None), color='green')
plt.legend(labels=['Fraud', 'Non-Fraud'])
plt.xlabel('Time')
```

The purpose of this scatter plot is to visually explore potential patterns or relationships between the time of fraudulent transactions and the normalized amount involved in those transactions. Such a visualization can be helpful to identify temporal trends, amount variations and outliers with extreme time or amount values that may require further investigation.

```python
sns.scatterplot(x='Time', y='Amount_Normalized' , data=df_creditCards.loc[df_creditCards['Class']==1])
plt.legend(labels=['Fraudulent Transaction'])
```

The box plots provide a visual representation of the distribution of normalized amount for each transaction class, allowing to identify  differences in central tendency, spread, and outlier behavior. These insights can be valuable for understanding fraud patterns and developing strategies for fraud detection and prevention.

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12,6))
#with outliers
s = sns.boxplot(ax = ax1, x="Class", y="Amount_Normalized", hue="Class",
                data=df_creditCards, palette="PRGn",showfliers=True)
#without outliers
s = sns.boxplot(ax = ax2, x="Class", y="Amount_Normalized", hue="Class",
                data=df_creditCards, palette="PRGn",showfliers=False)
```

```python
columns = df_creditCards.columns.values # get column/feature names

transactions_nonFraud = df_creditCards.loc[df_creditCards['Class'] == 0]
transactions_fraud = df_creditCards.loc[df_creditCards['Class'] == 1]

plt.figure()
fig, ax = plt.subplots(8,4,figsize=(16,28))

# Calculate variance for each feature in each class
transactions_nonFraud_variance = transactions_nonFraud.var()
transactions_fraud_variance = transactions_fraud.var()

# Filter out features with zero variance
featuresWithNonZeroVariance = transactions_nonFraud_variance[transactions_nonFraud_variance != 0].index.tolist()

for i, feature in enumerate(featuresWithNonZeroVariance):
    plt.subplot(8, 4, i + 1)
    sns.kdeplot(transactions_nonFraud[feature], bw_method=0.5, label="Class = 0")
    sns.kdeplot(transactions_fraud[feature], bw_method=0.5, label="Class = 1")
    plt.xlabel(feature, fontsize=12)
    plt.tick_params(axis='both', which='major', labelsize=12)

plt.show()
```

# Data Modeling

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
model = df_creditCards[['V1', 'V2', 'V3', 'V4', 'V5', 'V6', 'V7', 'V8', 'V9', 'V10',
       'V11', 'V12', 'V13', 'V14', 'V15', 'V16', 'V17', 'V20','V21', 'V22', 'V23', 'V25', 'V26', 'V27','Amount_Normalized','Class']]
       # Time is not necessary for predictions and increases performance cost

x = model.iloc[:,:-1] # feature data
y = model.iloc[:,-1] # target variable data

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, stratify=y, random_state=42)
# a certain seed is used for demonstration purposes
# 'stratify' is crucial for the proportion of fraudulent/non-fraudulent transactions,
# maintaining in both training and testing sets. It is essential for datasets that have imbalance.
```

```python
logreg = LogisticRegression()
logreg.fit(X_train, y_train)
# Identifying the relationship between fraudulent and non-fraudulent transactions
y_pred = logreg.predict(X_test)
```

# Model Evaluation

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
conf_matrix = pd.DataFrame(data=cm, columns=['Predicted:0', 'Predicted:1'], index=['Actual:0', 'Actual:1'])
plt.figure(figsize = (8,5))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap="YlGnBu");
```

```python
from sklearn.preprocessing import binarize
for i in range(0,11):
    cm2 = 0
    y_pred_prob_active = logreg.predict_proba(X_test)
    y_pred2 = binarize(y_pred_prob_active, threshold=i/10)[:,1]
    cm2 = confusion_matrix(y_test,y_pred2)
    print ('The confusion matrix is with', i/10, 'threshold is ', '\n', cm2, '\n',
            'with', cm2[0,0]+cm2[1,1], 'correct predictions and', cm2[1,0], 'False Negatives', '\n\n',
          'Sensitivity: ', cm2[1,1]/(float(cm2[1,1]+cm2[1,0])), 'Specificity: ', cm2[0,0]/(float(cm2[0,0]+cm2[0,1])), '\n\n\n')
```

```python
from sklearn.metrics import roc_curve
fpr, tpr, thresholds = roc_curve(y_test, y_pred_prob_active[:,1])
plt.plot(fpr,tpr)
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.0])
plt.title('ROC curve')
plt.xlabel('False positive rate') # 1-Specificity
plt.ylabel('True positive rate') # Sensitivity
plt.grid(True)
```

```python
from sklearn.metrics import roc_auc_score
roc_auc_score(y_test,y_pred_prob_active[:,1])
```
